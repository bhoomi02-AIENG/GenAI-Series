PROMPTS TEMPLATES - DYNAMIC PROMTS|

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
```

```python
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash-lite")
llm_2=ChatGroq(model="llama-3.1-8b-instant")
out=StrOutputParser()
```

```python
# MY SELF CHAIN NODE
def transform_case(result:str):
    return result.upper()
```

```python
prompts=ChatPromptTemplate.from_messages([
    {"role":"system","content":"You are a translator and translator input in {language} ?"},
    {"role":"user","content":"{query}"},

])
prompts
```

```python
final_prompts=prompts.format_messages(
     language="Hindi",
    query="I love python and javascripts"
)
final_prompts
```

```python
final_prompts=prompts.format_messages(
     language="Hindi",
    query="I love python and javascripts"
)

res=llm_2.invoke(final_prompts)
print(res.content)
```

```python
final_prompts=prompts.format_messages(
     language="Hinglish",
    query="I love python and javascripts"
)

res=llm_2.invoke(final_prompts)
print(res.content)
```

CHAINS RUNNABLE


```python
res=llm_2.invoke(prompts.invoke({"language":"hindi","query":"Python is coding language?"}))
res.content
```

```python
chains= prompts| llm_2 | out | transform_case
```

```python
chains
```

```python
res=chains.invoke({"language":"hindi", "query":"Python is coding language?"})
print(res)
```
