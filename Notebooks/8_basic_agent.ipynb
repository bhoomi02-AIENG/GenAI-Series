```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a:int, b:int):
    """
    It will return the sum of two numbers

    Args:
        a : Number One
        b : Number Two

    """
    return a+b
```

```python
add_numbers.invoke({"a":12, "b":8})
```

```python
@tool
def multiply_numbers(a:int, b:int):
    """
    It will return the product of two numbers

    Args:
        a : Number One
        b : Number Two

    """
    return a*b
```

```python
multiply_numbers.invoke({"a":12, "b":26})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
```

```python
llm=ChatGroq(model="llama-3.1-8b-instant")
```

```python
agent=create_agent(
    model=llm,
    tools=[add_numbers, multiply_numbers],
    system_prompt="You are a math teacher, and always use tool for calculation."
)
```

```python
res=agent.invoke({"messages":[{"role":"user", "content":"What is 2 + 6 ?"}]})
```

```python
res['messages']
```

```python
for resp in res['messages']:
    print(resp)
    print("\n")
```

```python
print(res['messages'][-1].content)
```
