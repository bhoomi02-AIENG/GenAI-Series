SOLUTION IS STRUCTURED OUTPUT

```python
from pydantic import BaseModel, Field  # pydantic helps to identify data types
from typing import List
class ResposeStrecture(BaseModel):
    name:str = Field(description="Complete Name")
    email:str = Field(description="Email Address")
    age:int = Field(description="Age")

strectured_llm=llm.with_structured_output(ResposeStrecture)
res=strectured_llm.invoke(f"Please give me only name, email and age from this text:{text}")
```

```python
res.name
```

```python
res.age
```

```python
res.model_dump()
```

```python
class Movies(BaseModel):
    title:str = Field(Description="Movie Title")
    year:int = Field(Description="Movie Released Year.")

movie_llm=llm.with_structured_output(Movies)
```

```python
movie_llm.invoke("Give me one recent movie")
```

```python
movie_llm.invoke("Give me one Sci-Fi movie")
```

```python
class AllMovies(BaseModel):
    movies:List[Movies]
movie_llm=llm.with_structured_output(AllMovies)    
```

```python
movie_llm.invoke("Give me three Sci-Fi movies")
```
