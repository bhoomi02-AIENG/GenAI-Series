```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm=ChatGroq(model="llama-3.1-8b-instant",streaming=True)
```

```python
question="What is GenAI?"
```

```python
res=llm.invoke(question)
res.content
```

```python
res=llm.stream(question)
res
```

```python
for chunk in res:
    print(chunk, "\n")
```
