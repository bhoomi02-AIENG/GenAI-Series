```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm_2=ChatGroq(model="llama-3.1-8b-instant")
```

```python
res=llm_2.invoke("Hello, My name is Bhumika.")
res.content
```

```python
res=llm_2.invoke("What is my name?")
res.content
```

```python
prompts=[
    {"role":"user","content":"Hello, Myself Bhumika."}
]
res=llm_2.invoke(prompts)
res.content

```

```python
prompts=[
    {"role":"user","content":"What is my name ?"}
]
res=llm_2.invoke(prompts)
res.content
```

```python
prompts=[
    {"role":"user","content":"Hello, Myself Bhumika."},
    {"role":"ai","content":"Hello Bhumika, it's nice to meet you. Is there something I can help you with or would you like to chat?"},
    {"role":"user", "content":"What is my name ?"},
]
res=llm_2.invoke(prompts)
res.content
```

MANAGE HISTORY

```python
history=[]
while True:
    query=input("User: ")
    if query.lower() in ["exit","quit","bye"]:
        print("Good bye👋")
        break
history.append({"role":"user", "content":query})
print("User: ",query)

res=llm_2.invoke(history)
history.append({"role":"ai","content":res.content})
print("AI: ",res.content,"\n")
```
